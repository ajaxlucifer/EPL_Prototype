Exploratory Data Analysis (EDA)
This notebook explores the dataset visually to identify patterns, distributions, and relationships between match statistics and match outcomes.


```python
#import and load data
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("epl_cleaned.csv")
data.head()
```

Target Variable Distribution

This section visualizes the distribution of home wins versus non-home wins to understand class balance.


```python
# bar chart
plt.figure(figsize=(6,4))
sns.countplot(x="home_win", data=data)
plt.title("Distribution of Match Outcomes")
plt.xlabel("Home Win (1 = Win, 0 = Not Win)")
plt.ylabel("Count")
plt.show()
```

Distribution of Match Statistics

Histograms are used to examine the distribution of key numerical features such as shots, fouls, and corners.


```python
#histogram
features = [
    "H Shots", "A Shots",
    "H SOT", "A SOT",
    "H Fouls", "A Fouls",
    "H Corners", "A Corners"
]

data[features].hist(figsize=(12,10), bins=20)
plt.tight_layout()
plt.show()
```

 Correlation Heatmap

This heatmap shows the correlation between numerical match statistics and the target variable.


```python
# heatmap
plt.figure(figsize=(12,8))
corr = data.corr(numeric_only=True)
sns.heatmap(corr, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap of Match Statistics")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="home_win", y="H Shots", data=data)
plt.title("Home Shots vs Match Outcome")
plt.show()
```

Key Insights from EDA

- Home teams generally record higher shots and shots on target.
- Match statistics such as shots on target and corners show noticeable correlation with match outcomes.
- The target variable is reasonably balanced, making accuracy and F1-score reliable evaluation metrics.
