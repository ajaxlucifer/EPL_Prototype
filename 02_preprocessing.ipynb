Data Preprocessing
This notebook focuses on feature selection and preprocessing for the Premier League match outcome prediction model.


```python
import pandas as pd

data = pd.read_csv("epl_cleaned.csv")
data.head()
```

Feature Selection

To avoid data leakage, features that directly describe the match outcome (such as full-time goals and result) are excluded.  
Only match statistics such as shots, fouls, corners, and disciplinary records are used as input features.


```python
#drop leakage columns
leakage_cols = [
    "FT Result", "FTH Goals", "FTA Goals",
    "HomeTeam", "AwayTeam",
    "Date", "Season", "League", "Display_Order"
]

data = data.drop(columns=leakage_cols, errors="ignore")
data.head()
```

```python
#defining x and y
X = data.drop(columns=["home_win"])
y = data["home_win"]

X.shape, y.shape
```

Preprocessing Summary

After removing leakage-prone and non-informative columns, the final feature matrix consists of numerical match statistics.  
These features are suitable for training classification models without introducing bias from the target variable.
